#### Chart - 7 Does the number reviews provided affect the number of bookings?

````python
# Chart - 6 visualization code
def iqr_outlier_removal(df):
    iqr = df.quantile(0.75) - df.quantile(0.25)
    lower_bound = df.quantile(0.25) - 1.5 * iqr
    upper_bound = df.quantile(0.75) + 1.5 * iqr
    df = df[((df >= lower_bound) & (df <= upper_bound))]
    return df
def percentile_outlier_removal(df1,lb,ub):
    lower_bound = lb
    upper_bound = ub
    q_lower = df1.quantile(lower_bound)
    q_upper = df1.quantile(upper_bound)
    
    ### use for single column of a dataframe as input
    df1 = df1[((df1 >= q_lower) & (df1 <= q_upper))]
    ### use for multiple columns of a dataframe as input
    #df1 = df1[((df1 >= q_lower) & (df1 <= q_upper)).all(axis=1)]
    return df1
def eyetest_outlier_removal(df1):
    age_mean = df1.mean()
    age_std = df1.std()
    age_threshold = age_mean + 2 * age_std

    #outliers_eye_test = df1[df1> age_threshold]
    df1 =df1[df1 < age_threshold ]
    return df1
````

````python
bookings['reviews_per_month'] = iqr_outlier_removal(bookings['reviews_per_month'])

print(bookings['reviews_per_month'].isnull().sum())
bookings  = bookings.query('reviews_per_month.notnull()')

print(bookings.shape)
print(bookings['reviews_per_month'].isnull().sum())
````
